```python

# Signed Binary Arithmetic using 16-bit 2's Complement

BITS = 16

def twos_complement(n, bits):
    """Convert signed decimal number to 2's complement binary."""
    if n < 0:
        n = (1 << bits) + n
    return format(n, f'0{bits}b')


def binary_to_decimal(binary):
    """Convert 2's complement binary to signed decimal."""
    value = int(binary, 2)

    if value >= (1 << (BITS - 1)):
        value -= (1 << BITS)

    return value


def add_binary(a, b):
    """Add two 16-bit binary numbers."""
    result = int(a, 2) + int(b, 2)
    return format(result & 0xFFFF, '016b')


def subtract_binary(a, b):
    """Subtract b from a using 2's complement."""
    # 2's complement of b
    b_complement = (~int(b, 2) + 1) & 0xFFFF

    result = int(a, 2) + b_complement

    return format(result & 0xFFFF, '016b')


def check_overflow(a, b, result, operation):
    """Check signed 16-bit arithmetic overflow."""
    if operation == '+':
        # Positive + Positive = Negative
        # Negative + Negative = Positive
        if (a >= 0 and b >= 0 and result < 0) or \
           (a < 0 and b < 0 and result >= 0):
            return True

    elif operation == '-':
        # Different signs can cause overflow
        if (a >= 0 and b < 0 and result < 0) or \
           (a < 0 and b >= 0 and result >= 0):
            return True

    return False


def perform_operation(a, b, operation):

    print("\nOperand A (8-bit):", twos_complement(a, 8))
    print("Operand B (8-bit):", twos_complement(b, 8))

    # Convert operands to 16-bit
    a16 = twos_complement(a, 16)
    b16 = twos_complement(b, 16)

    print("\n16-bit Operands:")
    print("A =", a16)
    print("B =", b16)

    if operation == '+':
        result_binary = add_binary(a16, b16)
    else:
        result_binary = subtract_binary(a16, b16)

    result_decimal = binary_to_decimal(result_binary)

    print("\nResult:")
    print("Binary  =", result_binary)
    print("Decimal =", result_decimal)

    if check_overflow(a, b, result_decimal, operation):
        print("Overflow = YES")
    else:
        print("Overflow = NO")


# (a) +56 + (+29)
print("Expression (a): +56 + (+29)")
perform_operation(56, 29, '+')


# (b) (-38) - (-23)
print("\nExpression (b): (-38) - (-23)")
perform_operation(-38, -23, '-')
```

```python

# Python program for ROR and RCR operations on 8-bit binary numbers

BITS = 8


def ROR(binary, positions):
    """Rotate Right without Carry"""
    print("\nROR Rotation Process:")

    for i in range(positions):
        last_bit = binary[-1]
        binary = last_bit + binary[:-1]
        print("After rotation", i + 1, ":", binary)

    return binary


def RCR(binary, carry, positions):
    """Rotate Right through Carry"""
    print("\nRCR Rotation Process:")

    for i in range(positions):
        last_bit = binary[-1]

        # Carry enters the MSB
        binary = str(carry) + binary[:-1]

        # Old LSB becomes the new carry
        carry = int(last_bit)

        print("After rotation", i + 1,
              ":", binary, " Carry =", carry)

    return binary, carry


# --------------------------------------------------
# (a) Rotate Right without Carry
# --------------------------------------------------

binary1 = "10111101"
positions = 3

print("======================================")
print("(a) ROR on", binary1, "by", positions, "positions")
print("======================================")

print("Original 8-bit number:", binary1)

result1 = ROR(binary1, positions)

print("Final ROR result:", result1)


# --------------------------------------------------
# (b) Rotate Right with Carry
# --------------------------------------------------

binary2 = "10100001"
carry = 0
positions = 3

print("\n======================================")
print("(b) RCR on", binary2, "by", positions, "positions")
print("======================================")

print("Original 8-bit number:", binary2)
print("Initial Carry:", carry)

result2, final_carry = RCR(binary2, carry, positions)

print("Final RCR result:", result2)
print("Final Carry:", final_carry)
```
